```python
# Scenario A: The "AI Support Router"
# Context: You are building a system that reads customer emails and decides which AI Agent should handle them .

# If the email contains the word "refund", route it to the Billing Agent.

# If it contains "password", route it to the Security Agent.

# Otherwise, route it to the General Support Agent.

# Data: emails = ["I forgot my password", "I want a refund please", "Hello, how are you?"]
```

```python
emails = ["I forgot my password", "I want a refund please", "Hello, how are you?"]

for email in emails:
    if "password" in email:
        print("Security Agent")
    elif "refund" in email:
        print("Billing Agent")
    else:
        print("General Support Agent")
```
